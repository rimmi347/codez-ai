```python
import matplotlib.pyplot as plt
from queue import Queue
import networkx as nx

adj_list = {
    "A": ["B", "D"],
    "B": ["A", "C"],
    "C": ["B"],
    "D": ["A", "E", "F"],
    "E": ["D", "F", "G"],
    "F": ["D", "E", "H"],
    "G": ["E", "H"],
    "H": ["G", "F"],
}

visited = {}
level = {}
parent = {}
bfs_traversal_output = []
queue = Queue()

for node in adj_list.keys():
    visited[node] = False
    parent[node] = None
    level[node] = -1

s = "A"
visited[s] = True
level[s] = 0
queue.put(s)


bfs_tree_edges = []

while not queue.empty():
    u = queue.get()
    bfs_traversal_output.append(u)

    for v in adj_list[u]:
        if not visited[v]:
            visited[v] = True
            parent[v] = u
            level[v] = level[u] + 1
            queue.put(v)
            bfs_tree_edges.append((u, v))

print("BFS Traversal Order:", bfs_traversal_output)
print("BFS Tree Edges:", bfs_tree_edges)


G = nx.Graph()
G.add_edges_from(bfs_tree_edges)

plt.figure(figsize=(6, 4))
nx.draw(
    G,
    with_labels=True,
    node_color="lightgreen",
    node_size=800,
    font_size=14,
    font_weight="bold",
    edge_color="black",
)
plt.title("BFS Tree Visualization")
plt.show()
```

```python
import matplotlib.pyplot as plt
import networkx as nx

Adj_list = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["B", "F"],
    "D": [],
    "E": ["F"],
    "F": [],
}

color = {}
parent = {}
trav_time = {}
dfs_traversal_output = []
dfs_tree_edges = []

for node in Adj_list.keys():
    color[node] = "W"
    parent[node] = None
    trav_time[node] = [-1, -1]

time = 0

def dfs_util(u):
    global time
    color[u] = "G"
    time += 1
    trav_time[u][0] = time
    dfs_traversal_output.append(u)

    for v in Adj_list[u]:
        if color[v] == "W":
            parent[v] = u
            dfs_tree_edges.append((u, v))
            dfs_util(v)

    color[u] = "B"
    time += 1
    trav_time[u][1] = time

for u in Adj_list.keys():
    if color[u] == "W":
        dfs_util(u)

print("DFS Traversal Output:", dfs_traversal_output)
print("Parent Mapping:", parent)
print("Traversal Times [Discovery, Finish]:", trav_time)
print("DFS Tree Edges:", dfs_tree_edges)


G = nx.DiGraph()
G.add_edges_from(dfs_tree_edges)

plt.figure(figsize=(6, 4))
nx.draw(
    G,
    with_labels=True,
    node_color="lightcoral",
    node_size=800,
    font_size=14,
    font_weight="bold",
    edge_color="black",
    arrowsize=20,
)
plt.title("DFS Tree Visualization")
plt.show()
```

```python
import matplotlib.pyplot as plt
import networkx as nx

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H', 'I'],
    'E': ['J', 'K'],
    'F': ['L', 'M'],
    'G': ['N', 'O'],
    'H': [],
    'I': [],
    'J': [],
    'K': [],
    'L': [],
    'M': [],
    'N': [],
    'O': [],
}


dls_edges = []

def DLS(start, goal, path, level, maxD):
    path.append(start)
    if start == goal:
        return path
    if level == maxD:
        return False
    for child in graph[start]:
        dls_edges.append((start, child))
        if DLS(child, goal, path, level + 1, maxD):
            return path
        path.pop()
    return False

start = 'A'
goal = 'K'
maxD = 3
path = []
res = DLS(start, goal, path, 0, maxD)

if res:
    print("Path to goal node available")
    print("Path:", path)
else:
    print("NO path available for the goal node in given depth limit")


G = nx.DiGraph()
G.add_edges_from(dls_edges)

plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G)
nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="lightblue",
    node_size=800,
    font_size=12,
    font_weight="bold",
    edge_color="black",
    arrows=True,
    arrowsize=26,
)
plt.title(f"DLS Tree Visualization (Depth Limit = {maxD})")
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

tree = {
    'J': ['A1', 'N1'],
    'A1': ['N2', 'A2'],
    'N1': ['T', 'U'],
    'N2': ['U2'],
    'A2': [],
    'T': [],
    'U': [],
    'U2': ['L'],
    'L': []
}

def dls(node, depth, visited):
    if depth < 0:
        return

    visited.append(node)

    if depth == 0:
        return

    for child in tree[node]:
        dls(child, depth - 1, visited)

def ids(start, max_depth):
    for depth in range(max_depth + 1):
        visited = []
        dls(start, depth, visited)
        print("Depth", depth, ":", visited)

ids('J', 4)

G = nx.DiGraph()

for parent, children in tree.items():
    for child in children:
        G.add_edge(parent, child)

labels = {
    'J': 'J',
    'A1': 'A',
    'N1': 'N',
    'N2': 'N',
    'A2': 'A',
    'T': 'T',
    'U': 'U',
    'U2': 'U',
    'L': 'L'
}

pos = {
    'J': (0, 4),
    'A1': (-1.5, 3),
    'N1': (1.5, 3),
    'N2': (-2, 2),
    'A2': (-1, 2),
    'T': (1, 2),
    'U': (2, 2),
    'U2': (-2, 1),
    'L': (-2, 0)
}

plt.figure(figsize=(9, 6))

nx.draw(
    G, pos,
    labels=labels,
    with_labels=True,
    node_color='violet',
    node_size=1800,
    font_color='white',
    font_size=14,
    font_weight='bold',
    arrows=True,
    arrowsize=20
)

plt.title("IDS Tree - JANNATUL")
plt.axis('off')
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import heapq

tree = {
    'J': [('A1', 2), ('N1', 4)],
    'A1': [('N2', 3), ('A2', 5)],
    'N1': [('T', 2), ('U1', 3)],
    'N2': [('A3', 2)],
    'A2': [],
    'T': [],
    'U1': [('U2', 2)],
    'A3': [('L', 3)],
    'U2': [],
    'L': []
}

labels = {
    'J': 'J',
    'A1': 'A',
    'N1': 'N',
    'N2': 'N',
    'A2': 'A',
    'T': 'T',
    'U1': 'U',
    'A3': 'A',
    'U2': 'U',
    'L': 'L'
}

def ucs(start, goal):
    queue = [(0, start, [start])]
    visited = set()

    while queue:
        cost, node, path = heapq.heappop(queue)

        if node in visited:
            continue

        visited.add(node)
        print("Visiting:", labels[node], "Cost:", cost)

        if node == goal:
            print("Goal found!")
            print("Path:", [labels[n] for n in path])
            print("Total Cost:", cost)
            return

        for child, edge_cost in tree[node]:
            if child not in visited:
                heapq.heappush(
                    queue,
                    (cost + edge_cost, child, path + [child])
                )

ucs('J', 'L')

G = nx.DiGraph()

for parent, children in tree.items():
    for child, cost in children:
        G.add_edge(parent, child, weight=cost)

pos = {
    'J': (0, 4),
    'A1': (-2, 3),
    'N1': (2, 3),
    'N2': (-3, 2),
    'A2': (-1, 2),
    'T': (1, 2),
    'U1': (3, 2),
    'A3': (-3, 1),
    'U2': (3, 1),
    'L': (-3, 0)
}

plt.figure(figsize=(10, 7))

nx.draw(
    G, pos,
    labels=labels,
    with_labels=True,
    node_color='violet',
    node_size=1800,
    font_color='white',
    font_size=14,
    font_weight='bold',
    arrows=True,
    arrowsize=20
)

edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(
    G, pos, edge_labels=edge_labels,
    font_size=11
)

plt.title("UCS Tree - JANNATUL")
plt.axis('off')
plt.show()
```
